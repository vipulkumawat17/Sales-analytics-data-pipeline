```python

import numpy as np

# Prepare order-level data
monthly_source = orders_df.copy()

monthly_source["order_purchase_timestamp"] = pd.to_datetime(
    monthly_source["order_purchase_timestamp"],
    errors="coerce"
)

monthly_source["order_month"] = (
    monthly_source["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype("string")
)

# Recalculate monthly metrics from order-level data
recalculated_monthly = (
    monthly_source
    .dropna(subset=["order_month"])
    .groupby("order_month", as_index=False)
    .agg(
        order_count=("order_id", "nunique"),
        item_revenue=("item_revenue", "sum"),
        freight_cost=("freight_cost", "sum"),
        total_items=("item_count", "sum"),
        orders_with_revenue=(
            "item_revenue",
            lambda x: (x > 0).sum()
        ),
        average_order_value=("item_revenue", "mean")
    )
)

# Load existing reference
monthly_reference = pd.read_csv(
    PROCESSED_DIR / "monthly_sales_analytics.csv"
)

monthly_reference["order_month"] = (
    pd.PeriodIndex(
        monthly_reference["order_month"],
        freq="M"
    ).astype(str)
)

# Compare metrics
comparison = recalculated_monthly.merge(
    monthly_reference,
    on="order_month",
    suffixes=("_recalculated", "_reference"),
    how="outer",
    indicator=True
)

metrics = [
    "order_count",
    "item_revenue",
    "freight_cost",
    "total_items",
    "orders_with_revenue",
    "average_order_value"
]

print("MONTHLY SALES RECONCILIATION")
print("=" * 65)

print("Months recalculated:", len(recalculated_monthly))
print("Months in reference:", len(monthly_reference))
print("Matching months:", (comparison["_merge"] == "both").sum())

for metric in metrics:
    left = comparison[f"{metric}_recalculated"]
    right = comparison[f"{metric}_reference"]

    matched = comparison["_merge"].eq("both")

    passed = (
        matched.all()
        and np.allclose(
            left[matched],
            right[matched],
            rtol=1e-5,
            atol=0.01,
            equal_nan=True
        )
    )

    print(f"{metric}: {'PASS' if passed else 'CHECK'}")

print("\nRecalculated monthly sample:")
display(recalculated_monthly.head(10))
```

```python

# 1. Identify unmatched months
print("UNMATCHED MONTHS")
print("=" * 60)

print(comparison.loc[
    comparison["_merge"] != "both",
    ["order_month", "_merge"]
].to_string(index=False))

# 2. Compare metrics only for common months
matched = comparison[comparison["_merge"] == "both"].copy()

print("\nMETRIC RECONCILIATION")
print("=" * 60)

for metric in metrics:
    left = matched[f"{metric}_recalculated"]
    right = matched[f"{metric}_reference"]

    differences = (left - right).abs()

    passed = np.allclose(
        left,
        right,
        rtol=1e-5,
        atol=0.01,
        equal_nan=True
    )

    print(f"\nMetric: {metric}")
    print("Status:", "PASS" if passed else "MISMATCH")
    print("Maximum absolute difference:", differences.max())

    if not passed:
        print(
            matched.loc[
                ~np.isclose(
                    left,
                    right,
                    rtol=1e-5,
                    atol=0.01,
                    equal_nan=True
                ),
                [
                    "order_month",
                    f"{metric}_recalculated",
                    f"{metric}_reference"
                ]
            ].head(10).to_string(index=False)
        )

# 3. Overall common-month result
all_passed = all(
    np.allclose(
        matched[f"{metric}_recalculated"],
        matched[f"{metric}_reference"],
        rtol=1e-5,
        atol=0.01,
        equal_nan=True
    )
    for metric in metrics
)

print("\n" + "=" * 60)
print("COMMON-MONTH RECONCILIATION:",
      "PASS" if all_passed else "MISMATCH")
print("Common months:", len(matched))
print("Recalculated months:", len(recalculated_monthly))
print("Reference months:", len(monthly_reference))
```

```python

# Test different order populations
base = orders_df.copy()

base["order_purchase_timestamp"] = pd.to_datetime(
    base["order_purchase_timestamp"],
    errors="coerce"
)

base["order_month"] = (
    base["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype("string")
)

# Candidate filtering conditions
candidates = {
    "All orders": base,
    "Orders with items": base[base["item_count"] > 0],
    "Orders with positive revenue": base[base["item_revenue"] > 0]
}

def calculate_monthly_metrics(df):
    return (
        df.dropna(subset=["order_month"])
        .groupby("order_month", as_index=False)
        .agg(
            order_count=("order_id", "nunique"),
            item_revenue=("item_revenue", "sum"),
            freight_cost=("freight_cost", "sum"),
            total_items=("item_count", "sum"),
            orders_with_revenue=(
                "item_revenue",
                lambda x: (x > 0).sum()
            ),
            average_order_value=("item_revenue", "mean")
        )
    )

print("POPULATION DIAGNOSTICS")
print("=" * 65)

for name, candidate in candidates.items():
    calculated = calculate_monthly_metrics(candidate)

    merged = calculated.merge(
        monthly_reference,
        on="order_month",
        suffixes=("_calc", "_ref"),
        how="inner"
    )

    results = {}

    for metric in metrics:
        results[metric] = np.allclose(
            merged[f"{metric}_calc"],
            merged[f"{metric}_ref"],
            rtol=1e-5,
            atol=0.01,
            equal_nan=True
        )

    print(f"\n{name}")
    print("Source rows:", len(candidate))
    print("Months:", len(calculated))
    print("Matching metrics:", sum(results.values()), "/", len(metrics))
    print("Metric results:", results)

# Investigate orders without items or revenue
print("\n\nORDER POPULATION SUMMARY")
print("=" * 65)

print("Total orders:", len(base))
print("Null item counts:", base["item_count"].isna().sum())
print("Zero item counts:", (base["item_count"] == 0).sum())
print("Positive item counts:", (base["item_count"] > 0).sum())
print("Null item revenue:", base["item_revenue"].isna().sum())
print("Zero item revenue:", (base["item_revenue"] == 0).sum())
print("Positive item revenue:", (base["item_revenue"] > 0).sum())

print("\nOCTOBER 2018 RECORDS")
october = base[base["order_month"] == "2018-10"]

print("Rows:", len(october))
display(
    october[
        [
            "order_id",
            "order_status",
            "item_count",
            "item_revenue",
            "freight_cost"
        ]
    ].head(10)
)
```

```python

# Reproduce original monthly transformation rules
EXCLUDED_SALES_STATUSES = {"canceled", "unavailable"}

eligible_orders = orders_df[
    ~orders_df["order_status"].isin(EXCLUDED_SALES_STATUSES)
].copy()

eligible_orders["order_purchase_timestamp"] = pd.to_datetime(
    eligible_orders["order_purchase_timestamp"],
    errors="coerce"
)

eligible_orders["order_month"] = (
    eligible_orders["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype("string")
)

# Match original aggregation definitions
monthly_reproduced = (
    eligible_orders
    .dropna(subset=["order_purchase_timestamp"])
    .groupby("order_month", as_index=False)
    .agg(
        order_count=("order_id", "nunique"),
        item_revenue=("item_revenue", "sum"),
        freight_cost=("freight_cost", "sum"),
        total_items=("item_count", "sum"),
        orders_with_revenue=("item_revenue", "count")
    )
)

monthly_reproduced["average_order_value"] = (
    monthly_reproduced["item_revenue"]
    / monthly_reproduced["orders_with_revenue"].replace(0, pd.NA)
)

monthly_reproduced = monthly_reproduced.sort_values(
    "order_month"
).reset_index(drop=True)

# Compare with original reference
monthly_comparison = monthly_reproduced.merge(
    monthly_reference,
    on="order_month",
    suffixes=("_new", "_original"),
    how="outer",
    indicator=True
)

print("FINAL MONTHLY RECONCILIATION")
print("=" * 65)

print("Reproduced months:", len(monthly_reproduced))
print("Reference months:", len(monthly_reference))
print("Matching months:", (monthly_comparison["_merge"] == "both").sum())

reconciliation_results = {}

for metric in metrics:
    left = monthly_comparison[f"{metric}_new"]
    right = monthly_comparison[f"{metric}_original"]

    common = monthly_comparison["_merge"] == "both"

    passed = (
        common.all()
        and np.allclose(
            left[common],
            right[common],
            rtol=1e-5,
            atol=0.01,
            equal_nan=True
        )
    )

    reconciliation_results[metric] = passed
    print(f"{metric}: {'PASS' if passed else 'FAIL'}")

print("\nOVERALL RESULT:",
      "PASS" if all(reconciliation_results.values()) else "FAIL")

print("\nReproduced monthly sample:")
display(monthly_reproduced.head(5))
```

Generate product and customer rankings

```python

# 1. Product rankings
product_rankings = products_df.copy()

product_rankings = product_rankings.sort_values(
    ["item_revenue", "units_sold", "product_id"],
    ascending=[False, False, True]
).reset_index(drop=True)

product_rankings["revenue_rank"] = range(
    1, len(product_rankings) + 1
)

product_rankings = product_rankings.sort_values(
    ["units_sold", "item_revenue", "product_id"],
    ascending=[False, False, True]
).reset_index(drop=True)

product_rankings["units_sold_rank"] = range(
    1, len(product_rankings) + 1
)

# 2. Customer rankings
customer_rankings = customers_df.copy()

customer_rankings = customer_rankings.sort_values(
    ["item_revenue", "order_count", "customer_unique_id"],
    ascending=[False, False, True]
).reset_index(drop=True)

customer_rankings["revenue_rank"] = range(
    1, len(customer_rankings) + 1
)

customer_rankings = customer_rankings.sort_values(
    ["order_count", "item_revenue", "customer_unique_id"],
    ascending=[False, False, True]
).reset_index(drop=True)

customer_rankings["order_count_rank"] = range(
    1, len(customer_rankings) + 1
)

# 3. Revenue contribution
total_product_revenue = product_rankings["item_revenue"].sum()

product_rankings["revenue_contribution_pct"] = (
    product_rankings["item_revenue"]
    / total_product_revenue * 100
)

# 4. Display top 10 results
print("TOP 10 PRODUCTS BY REVENUE")
print("=" * 65)

display(
    product_rankings.sort_values("revenue_rank").head(10)[
        [
            "revenue_rank",
            "product_id",
            "product_category_name_english",
            "units_sold",
            "item_revenue",
            "distinct_orders",
            "revenue_contribution_pct"
        ]
    ]
)

print("\nTOP 10 CUSTOMERS BY REVENUE")
print("=" * 65)

display(
    customer_rankings.sort_values("revenue_rank").head(10)[
        [
            "revenue_rank",
            "customer_unique_id",
            "order_count",
            "item_revenue",
            "average_order_value",
            "customer_state"
        ]
    ]
)

print("\nTOP 10 CUSTOMERS BY ORDER FREQUENCY")
print("=" * 65)

display(
    customer_rankings.sort_values("order_count_rank").head(10)[
        [
            "order_count_rank",
            "customer_unique_id",
            "order_count",
            "item_revenue",
            "is_repeat_customer"
        ]
    ]
)

print("\nRanking datasets generated:")
print("Products:", len(product_rankings))
print("Customers:", len(customer_rankings))
```

```python
import pandas as pd

# Load only the raw tables required for quality-rate calculations
raw = {
    "customers": pd.read_csv(DATA_DIR / "olist_customers_dataset.csv"),
    "orders": pd.read_csv(DATA_DIR / "olist_orders_dataset.csv"),
    "items": pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv"),
    "products": pd.read_csv(DATA_DIR / "olist_products_dataset.csv"),
    "sellers": pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv"),
    "payments": pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv"),
    "reviews": pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv"),
}

# 1. Required-field null rate
required_fields = [
    ("customers", "customer_id"),
    ("orders", "order_id"),
    ("items", "order_id"),
    ("items", "order_item_id"),
    ("products", "product_id"),
    ("sellers", "seller_id"),
    ("orders", "customer_id"),
    ("items", "product_id"),
    ("items", "seller_id"),
    ("payments", "order_id"),
    ("reviews", "order_id"),
]

null_count = sum(raw[t][c].isna().sum() for t, c in required_fields)
required_value_count = sum(len(raw[t]) for t, c in required_fields)

# 2. Duplicate primary/composite-key rate
unique_keys = [
    ("customers", ["customer_id"]),
    ("orders", ["order_id"]),
    ("items", ["order_id", "order_item_id"]),
    ("products", ["product_id"]),
    ("sellers", ["seller_id"]),
]

duplicate_count = sum(
    raw[t].duplicated(subset=keys).sum()
    for t, keys in unique_keys
)
unique_key_rows = sum(len(raw[t]) for t, keys in unique_keys)

# 3. Referential-integrity violation rate
foreign_keys = [
    ("orders", "customer_id", "customers", "customer_id"),
    ("items", "product_id", "products", "product_id"),
    ("items", "seller_id", "sellers", "seller_id"),
    ("items", "order_id", "orders", "order_id"),
    ("payments", "order_id", "orders", "order_id"),
    ("reviews", "order_id", "orders", "order_id"),
]

fk_violations = 0
fk_values_checked = 0

for child_table, child_col, parent_table, parent_col in foreign_keys:
    child_values = raw[child_table][child_col].dropna()
    parent_values = raw[parent_table][parent_col].dropna()

    fk_violations += (~child_values.isin(parent_values)).sum()
    fk_values_checked += len(child_values)

# 4. Invalid numeric-value rate
numeric_fields = [
    ("items", "price"),
    ("items", "freight_value"),
    ("payments", "payment_value"),
    ("payments", "payment_installments"),
]

invalid_numeric_count = 0
numeric_values_checked = 0

for table, column in numeric_fields:
    original = raw[table][column]
    numeric = pd.to_numeric(original, errors="coerce")
    non_null = original.notna()

    invalid = non_null & (numeric.isna() | (numeric < 0))

    invalid_numeric_count += invalid.sum()
    numeric_values_checked += non_null.sum()

# 5. Invalid review-score rate
scores_original = raw["reviews"]["review_score"]
scores = pd.to_numeric(scores_original, errors="coerce")

invalid_scores = (
    scores.isna()
    | (scores % 1 != 0)
    | (scores < 1)
    | (scores > 5)
)

invalid_score_count = invalid_scores.sum()
review_score_count = len(scores)

# 6. Overall validation check rate
total_checks = len(validation_results)
passed_checks = sum(
    result["status"] == "PASS"
    for result in validation_results
)

# Build evaluation metrics
quality_metrics = pd.DataFrame([
    {
        "metric": "validation_check_pass_rate",
        "numerator": passed_checks,
        "denominator": total_checks,
        "rate_pct": passed_checks / total_checks * 100,
        "definition": "Passed validation checks divided by total validation checks",
    },
    {
        "metric": "required_field_null_rate",
        "numerator": null_count,
        "denominator": required_value_count,
        "rate_pct": null_count / required_value_count * 100,
        "definition": "Null values across required fields divided by all required-field values examined",
    },
    {
        "metric": "duplicate_key_rate",
        "numerator": duplicate_count,
        "denominator": unique_key_rows,
        "rate_pct": duplicate_count / unique_key_rows * 100,
        "definition": "Duplicate key rows across tested primary/composite keys divided by rows examined",
    },
    {
        "metric": "referential_integrity_violation_rate",
        "numerator": fk_violations,
        "denominator": fk_values_checked,
        "rate_pct": fk_violations / fk_values_checked * 100,
        "definition": "Unmatched non-null foreign-key values divided by non-null foreign-key values examined",
    },
    {
        "metric": "invalid_numeric_value_rate",
        "numerator": invalid_numeric_count,
        "denominator": numeric_values_checked,
        "rate_pct": invalid_numeric_count / numeric_values_checked * 100,
        "definition": "Negative or unparseable non-null numeric values divided by non-null numeric values examined",
    },
    {
        "metric": "invalid_review_score_rate",
        "numerator": invalid_score_count,
        "denominator": review_score_count,
        "rate_pct": invalid_score_count / review_score_count * 100,
        "definition": "Null, non-integer, or out-of-range review scores divided by all review rows",
    },
])

# Save evaluation output
quality_metrics.to_csv(
    PHASE6_DIR / "data_quality_metrics.csv",
    index=False
)

print("DATA QUALITY METRICS")
print(quality_metrics.to_string(index=False))
print("\nSaved:", PHASE6_DIR / "data_quality_metrics.csv")
```

```python
import time
import statistics
import pandas as pd
import numpy as np

from pyspark.sql import SparkSession, functions as F

# Reuse the existing Spark session
if "spark" not in globals():
    spark = (
        SparkSession.builder
        .appName("SalesAnalyticsPhase6Benchmark")
        .master("local[*]")
        .config("spark.sql.shuffle.partitions", "4")
        .getOrCreate()
    )

ITEMS_PATH = DATA_DIR / "olist_order_items_dataset.csv"

# Load the same source data before timing
pandas_items = pd.read_csv(ITEMS_PATH)

spark_items = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(str(ITEMS_PATH))
    .cache()
)

spark_input_rows = spark_items.count()

print("Input rows — Pandas:", len(pandas_items))
print("Input rows — PySpark:", spark_input_rows)

# Equivalent transformations
def pandas_order_item_aggregation(df):
    return (
        df.groupby("order_id", as_index=False)
        .agg(
            item_revenue=("price", "sum"),
            freight_cost=("freight_value", "sum"),
            item_count=("order_item_id", "count"),
        )
        .sort_values("order_id")
        .reset_index(drop=True)
    )


def spark_order_item_aggregation(df):
    return (
        df.groupBy("order_id")
        .agg(
            F.sum("price").alias("item_revenue"),
            F.sum("freight_value").alias("freight_cost"),
            F.count("order_item_id").alias("item_count"),
        )
    )


# Warm-up runs — excluded from measured timings
_ = pandas_order_item_aggregation(pandas_items)

spark_warmup = spark_order_item_aggregation(spark_items)
spark_warmup.agg(
    F.sum("item_count"),
    F.sum("item_revenue"),
    F.sum("freight_cost"),
).collect()

# Benchmark three executions
pandas_times = []
spark_times = []

for run_number in range(3):
    start = time.perf_counter()
    pandas_result = pandas_order_item_aggregation(pandas_items)
    pandas_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    spark_result = spark_order_item_aggregation(spark_items)

    # Force execution of the grouped aggregation
    spark_result.agg(
        F.sum("item_count"),
        F.sum("item_revenue"),
        F.sum("freight_cost"),
    ).collect()

    spark_times.append(time.perf_counter() - start)

# Reconcile outputs outside the timed benchmark
spark_result_pd = (
    spark_result
    .select("order_id", "item_revenue", "freight_cost", "item_count")
    .orderBy("order_id")
    .toPandas()
    .reset_index(drop=True)
)

pandas_result = pandas_result[
    ["order_id", "item_revenue", "freight_cost", "item_count"]
].reset_index(drop=True)

assert len(pandas_result) == len(spark_result_pd), (
    "Output row counts do not match"
)

assert pandas_result["order_id"].equals(spark_result_pd["order_id"]), (
    "Order IDs do not match"
)

for column in ["item_revenue", "freight_cost"]:
    assert np.allclose(
        pandas_result[column],
        spark_result_pd[column],
        rtol=1e-9,
        atol=1e-6,
        equal_nan=True,
    ), f"{column} reconciliation failed"

assert np.array_equal(
    pandas_result["item_count"],
    spark_result_pd["item_count"],
), "Item counts do not match"

# Summarize benchmark results
pandas_median = statistics.median(pandas_times)
spark_median = statistics.median(spark_times)

benchmark_results = pd.DataFrame([
    {
        "engine": "Pandas",
        "input_rows": len(pandas_items),
        "output_rows": len(pandas_result),
        "runs": len(pandas_times),
        "median_execution_seconds": pandas_median,
        "min_execution_seconds": min(pandas_times),
        "max_execution_seconds": max(pandas_times),
    },
    {
        "engine": "PySpark",
        "input_rows": spark_input_rows,
        "output_rows": len(spark_result_pd),
        "runs": len(spark_times),
        "median_execution_seconds": spark_median,
        "min_execution_seconds": min(spark_times),
        "max_execution_seconds": max(spark_times),
    },
])

benchmark_results["execution_speedup_vs_pandas"] = [
    1.0,
    pandas_median / spark_median if spark_median else np.nan,
]

# Save only after successful reconciliation
benchmark_path = PHASE6_DIR / "performance_benchmark.csv"
benchmark_results.to_csv(benchmark_path, index=False)

print("\nPERFORMANCE BENCHMARK")
print(benchmark_results.to_string(index=False))

print("\nRECONCILIATION")
print("Pandas output rows:", len(pandas_result))
print("PySpark output rows:", len(spark_result_pd))
print("Item count: PASS")
print("Item revenue: PASS")
print("Freight cost: PASS")
print("Overall reconciliation: PASS")

print("\nSaved:", benchmark_path)

spark_items.unpersist()
```

```python
from pathlib import Path
import subprocess

phase6_files = [
    "reports/phase6/daily_sales_metrics.csv",
    "reports/phase6/monthly_sales_metrics.csv",
    "reports/phase6/product_rankings.csv",
    "reports/phase6/customer_rankings.csv",
    "reports/phase6/data_quality_metrics.csv",
    "reports/phase6/performance_benchmark.csv",
    "reports/phase6/phase6_findings.md",
    "reports/phase6/charts/daily_sales_trend.png",
    "reports/phase6/charts/monthly_sales_trend.png",
    "reports/phase6/charts/top_10_products_by_revenue.png",
    "reports/phase6/charts/top_10_customers_by_revenue.png",
    "reports/phase6/charts/data_quality_rates.png",
    "reports/phase6/charts/performance_comparison.png",
]

for relative_path in phase6_files:
    path = REPO / relative_path
    exists = path.exists()

    check = subprocess.run(
        ["git", "check-ignore", "-v", relative_path],
        cwd=REPO,
        capture_output=True,
        text=True,
    )

    tracked = subprocess.run(
        ["git", "ls-files", "--error-unmatch", relative_path],
        cwd=REPO,
        capture_output=True,
        text=True,
    ).returncode == 0

    if not exists:
        state = "MISSING"
    elif tracked:
        state = "TRACKED"
    elif check.returncode == 0:
        state = "IGNORED"
    else:
        state = "UNTRACKED"

    print(f"{state:10} {relative_path}")
    if check.returncode == 0:
        print(" " * 11 + check.stdout.strip())
```
