```python
from pathlib import Path
import subprocess

# Verify Google Drive
assert PROJECT_ROOT.exists(), "Google Drive project folder not found."

# Verify Git repository
result = subprocess.run(
    ["git", "rev-parse", "--is-inside-work-tree"],
    cwd=REPO_PATH,
    capture_output=True,
    text=True,
    check=True,
)

print("Google Drive: Connected")
print("Git repository:", result.stdout.strip())

# Show current branch and pending changes
subprocess.run(["git", "branch", "--show-current"], cwd=REPO_PATH)
subprocess.run(["git", "status", "--short"], cwd=REPO_PATH)
```

```python
def git(*args):
    return subprocess.run(
        ["git", *args],
        cwd=str(REPO_PATH),
        capture_output=True,
        text=True,
        check=True,
    ).stdout.strip()

print("Repository:", REPO_PATH)
print("Branch:", git("branch", "--show-current"))
print("Remote:")
print(git("remote", "-v"))
print("Working tree:")
print(git("status", "--short"))
```

```python
from pathlib import Path

folders = [
    "src/sales_pipeline",
    "tests",
    "notebooks",
    "docs",
    "config",
]

for folder in folders:
  (REPO_PATH / folder).mkdir(parents=True, exist_ok=True)

#Keep otherwise-empty directories visible in Git
for folder in [
    "src/sales_pipeline",
    "tests",
    "notebooks",
    "docs",
    "config"
]:

  keep_file = REPO_PATH/ folder / ".gitkeep"
  keep_file.touch(exist_ok=True)

print("Project Directories created:")
for folder in folders:
  print(f"- {folder}/")
```
