```python
import subprocess
from pathlib import Path

REPO_PATH = Path(
    "/content/drive/MyDrive/Sales-analytics-data-pipeline/"
    "Sales-analytics-data-pipeline"
)

assert (REPO_PATH / ".git").is_dir(), (
    f"Repository not found: {REPO_PATH}"
)

def check_git(title, *args):
    print(f"\n--- {title} ---")
    result = subprocess.run(
        ["git", *args],
        cwd=str(REPO_PATH),
        capture_output=True,
        text=True,
    )
    print(result.stdout.strip() or "(No output)")
    if result.stderr.strip():
        print("Details:", result.stderr.strip())
    print("Exit code:", result.returncode)

check_git("Global username", "config", "--global", "--get", "user.name")
check_git("Global email", "config", "--global", "--get", "user.email")

check_git("Local username", "config", "--local", "--get", "user.name")
check_git("Local email", "config", "--local", "--get", "user.email")

check_git("Latest commit author", "log", "-1", "--format=%an <%ae>")
check_git("Latest commit", "log", "-1", "--oneline", "--decorate")
check_git("Branch status", "status", "-sb")
check_git("Remote configuration", "remote", "-v")
```
